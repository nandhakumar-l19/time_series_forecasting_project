```python


import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error
import math
import tensorflow as tf
from tensorflow.keras import layers, models, optimizers, callbacks
from tqdm import tqdm



KAGGLE_DATASET_SLUG = "uciml/electric-power-consumption-data-set"
DOWNLOAD_DIR = "./data"
CSV_FILENAME = "household_power_consumption.txt"

LOOKBACK = 168
HORIZON = 24
BATCH_SIZE = 64
EPOCHS = 20
TEST_SPLIT_RATIO = 0.1
VAL_SPLIT_RATIO = 0.1
RANDOM_SEED = 42

os.makedirs(DOWNLOAD_DIR, exist_ok=True)
np.random.seed(RANDOM_SEED)
tf.random.set_seed(RANDOM_SEED)


def download_kaggle_dataset(dataset_slug: str, dest: str):
    """Uses kaggle CLI to download a dataset (requires kaggle.json credentials)."""
    print("Attempting Kaggle download of:", dataset_slug)

    cmd = f"kaggle datasets download -d {dataset_slug} -p {dest} --unzip"
    ret = os.system(cmd)
    if ret != 0:
        print("Kaggle download failed — maybe credentials missing. Please download manually from Kaggle.")
    else:
        print("Download complete.")

download_kaggle_dataset(KAGGLE_DATASET_SLUG, DOWNLOAD_DIR)


def load_household_power(path):
    """
    Expected file: 'household_power_consumption.txt' (from UCI / Kaggle)
    Format: Date;Time;Global_active_power;Global_reactive_power;Voltage;Global_intensity;Sub_metering_1;Sub_metering_2;Sub_metering_3
    """
    fpath = os.path.join(path, CSV_FILENAME)
    if not os.path.exists(fpath):
        raise FileNotFoundError(f"{fpath} not found. Place dataset file here or change CSV_FILENAME.")

    df = pd.read_csv(fpath, sep=';', low_memory=False, na_values=['?'])

    df['datetime'] = pd.to_datetime(df['Date'] + ' ' + df['Time'], dayfirst=True, errors='coerce')
    df.drop(columns=['Date', 'Time'], inplace=True)
    df.set_index('datetime', inplace=True)

    for col in df.columns:
        df[col] = pd.to_numeric(df[col], errors='coerce')
    return df

print("Loading dataset...")
try:
    df_raw = load_household_power(DOWNLOAD_DIR)
except FileNotFoundError as e:
    print(e)
    print("To proceed for demo, generating synthetic multivariate time series instead.")

    rng = pd.date_range("2017-01-01", periods=20000, freq='H')
    df_raw = pd.DataFrame({
        'global_active_power': 1.0 + 0.5*np.sin(np.arange(len(rng))/24*2*np.pi) + 0.1*np.random.randn(len(rng)),
        'voltage': 230 + 5*np.sin(np.arange(len(rng))/24*2*np.pi*0.5) + 0.5*np.random.randn(len(rng)),
        'sub_metering_1': np.abs(0.2*np.sin(np.arange(len(rng))/24*2*np.pi*3) + 0.05*np.random.randn(len(rng))),
    }, index=rng)

print("Raw data shape:", df_raw.shape)
print("Columns:", df_raw.columns.tolist())


def preprocess_timeseries(df, resample_rule='H'):

    df = df.copy()

    df = df.resample(resample_rule).mean()

    df = df.interpolate(method='time').ffill().bfill()
    return df

df = preprocess_timeseries(df_raw, resample_rule='H')
print("Preprocessed data shape:", df.shape)


if len(df) < 2000:
    print("WARNING: dataset <2000 timesteps; some experiments expect >=1000. Consider a longer dataset.")


FEATURE_COLS = df.columns.tolist()
TARGET_COL = 'global_active_power' if 'global_active_power' in df.columns else df.columns[0]


n_total = len(df)
n_test = int(n_total * TEST_SPLIT_RATIO)
n_val = int(n_total * VAL_SPLIT_RATIO)
n_train = n_total - n_test - n_val

df_train = df.iloc[:n_train]
df_val = df.iloc[n_train:n_train+n_val]
df_test = df.iloc[n_train+n_val:]

scaler = StandardScaler()
scaler.fit(df_train.values)
df_scaled = pd.DataFrame(scaler.transform(df), index=df.index, columns=df.columns)
df_train_s = df_scaled.iloc[:n_train]
df_val_s = df_scaled.iloc[n_train:n_train+n_val]
df_test_s = df_scaled.iloc[n_train+n_val:]

print("Train/Val/Test sizes:", len(df_train_s), len(df_val_s), len(df_test_s))


def create_windows(data_df, lookback=LOOKBACK, horizon=HORIZON, feature_cols=None, target_col=None, step=1):
    X, Y = [], []
    arr = data_df[feature_cols].values if feature_cols else data_df.values
    target_idx = data_df.columns.get_loc(target_col) if target_col else 0
    for start in range(0, len(arr) - lookback - horizon + 1, step):
        x = arr[start:start+lookback]
        y = arr[start+lookback:start+lookback+horizon, target_idx]
        X.append(x)
        Y.append(y)
    X = np.array(X)
    Y = np.array(Y)
    return X, Y

feat_cols = FEATURE_COLS
X_train, y_train = create_windows(df_train_s, LOOKBACK, HORIZON, feat_cols, TARGET_COL)
X_val, y_val = create_windows(pd.concat([df_train_s.tail(LOOKBACK), df_val_s]), LOOKBACK, HORIZON, feat_cols, TARGET_COL)
X_test, y_test = create_windows(pd.concat([df_val_s.tail(LOOKBACK), df_test_s]), LOOKBACK, HORIZON, feat_cols, TARGET_COL)

print("Window shapes:", X_train.shape, y_train.shape, X_val.shape, y_val.shape, X_test.shape, y_test.shape)

def mape(y_true, y_pred):
    return np.mean(np.abs((y_true - y_pred) / (np.where(y_true==0, 1e-8, y_true)))) * 100

def rmse(y_true, y_pred):
    return np.sqrt(mean_squared_error(y_true, y_pred))


def build_lstm_model(input_shape, horizon):
    inp = layers.Input(shape=input_shape)
    x = layers.Masking()(inp)
    x = layers.LSTM(128, return_sequences=False)(x)
    x = layers.Dropout(0.2)(x)
    out = layers.Dense(horizon)(x)
    model = models.Model(inp, out)
    model.compile(optimizer=optimizers.Adam(1e-3), loss='mse', metrics=['mae'])
    return model

lstm = build_lstm_model((LOOKBACK, len(feat_cols)), HORIZON)
lstm.summary()

es = callbacks.EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)
history_lstm = lstm.fit(
    X_train, y_train,
    validation_data=(X_val, y_val),
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    callbacks=[es],
    verbose=2
)


y_pred_lstm = lstm.predict(X_test)
print("LSTM Test MAE:", mean_absolute_error(y_test.flatten(), y_pred_lstm.flatten()))
print("LSTM Test RMSE:", rmse(y_test.flatten(), y_pred_lstm.flatten()))
print("LSTM Test MAPE:", mape(y_test, y_pred_lstm))

class TransformerEncoderLayer(layers.Layer):
    def __init__(self, d_model, num_heads, dff, dropout=0.1, **kwargs):
        super().__init__(**kwargs)
        self.mha = layers.MultiHeadAttention(num_heads=num_heads, key_dim=d_model//num_heads, dropout=dropout)
        self.ffn = tf.keras.Sequential([
            layers.Dense(dff, activation='relu'),
            layers.Dense(d_model),
        ])
        self.layernorm1 = layers.LayerNormalization(epsilon=1e-6)
        self.layernorm2 = layers.LayerNormalization(epsilon=1e-6)
        self.dropout1 = layers.Dropout(dropout)
        self.dropout2 = layers.Dropout(dropout)

    def call(self, x, training, return_attention_scores=False):

        if return_attention_scores:
            attn_out, attn_scores = self.mha(x, x, return_attention_scores=True, training=training)
        else:
            attn_out = self.mha(x, x, training=training)
            attn_scores = None
        attn_out = self.dropout1(attn_out, training=training)
        out1 = self.layernorm1(x + attn_out)
        ffn_out = self.ffn(out1)
        ffn_out = self.dropout2(ffn_out, training=training)
        out2 = self.layernorm2(out1 + ffn_out)
        return out2, attn_scores

def build_transformer_forecaster(input_shape, horizon, d_model=64, num_heads=4, dff=128, n_layers=2):
    inp = layers.Input(shape=input_shape, name='encoder_input')
    x = inp

    x = layers.Dense(d_model)(x)

    pos = layers.Embedding(input_dim=input_shape[0], output_dim=d_model)
    positions = tf.range(start=0, limit=input_shape[0], delta=1)
    pos_enc = pos(positions)
    x = x + pos_enc

    attn_scores_all = []
    for i in range(n_layers):
        layer = TransformerEncoderLayer(d_model, num_heads, dff, name=f"encoder_layer_{i}")
        x, attn_scores = layer(x, training=True, return_attention_scores=True)
        attn_scores_all.append(attn_scores)
    x = layers.GlobalAveragePooling1D()(x)
    x = layers.Dense(128, activation='relu')(x)
    out = layers.Dense(horizon)(x)
    model = models.Model(inp, out)
    model.compile(optimizer=optimizers.Adam(1e-3), loss='mse', metrics=['mae'])

    return model

transformer = build_transformer_forecaster((LOOKBACK, len(feat_cols)), HORIZON)
transformer.summary()

history_tf = transformer.fit(
    X_train, y_train,
    validation_data=(X_val, y_val),
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    callbacks=[es],
    verbose=2
)

y_pred_tf = transformer.predict(X_test)
print("Transformer Test MAE:", mean_absolute_error(y_test.flatten(), y_pred_tf.flatten()))
print("Transformer Test RMSE:", rmse(y_test.flatten(), y_pred_tf.flatten()))
print("Transformer Test MAPE:", mape(y_test, y_pred_tf))


def build_attention_extractor(base_model, input_shape):
    """Return a model that takes same input and returns the outputs of MHA attention scores from encoder layers."""

    inputs = base_model.input

    x = inputs
    x = base_model.get_layer(index=2).input

    d_model = base_model.get_layer(index=2).output_shape[-1] if len(base_model.layers) > 2 else 64

    raise NotImplementedError("Attention extraction code path requires building model with explicit access to encoder layers. "
                              "In this script the model returned attention scores during training but Keras saved opaque layers. "
                              "See comments in notebook for how to set up extractor with layer references.")


print("\nTo extract attention scores reliably: build the Transformer by keeping references to encoder layers inside a wrapper class "
      "or re-create the encoder using the saved weights and call each layer with return_attention_scores=True. "
      "This demo shows where to hook attention scores and how to visualize.\n")


def plot_predictions(y_true, y_pred, n_plots=3):
    for i in range(n_plots):
        plt.figure(figsize=(8,2.5))
        plt.plot(y_true[i], label='true')
        plt.plot(y_pred[i], label='pred')
        plt.title(f"Sample {i} — MAE={mean_absolute_error(y_true[i], y_pred[i]):.4f}")
        plt.legend()
        plt.show()

plot_predictions(y_test, y_pred_tf, n_plots=3)




MODEL_DIR = "./models"
os.makedirs(MODEL_DIR, exist_ok=True)
lstm.save(os.path.join(MODEL_DIR, "lstm_baseline.h5"))
transformer.save(os.path.join(MODEL_DIR, "transformer_forecast.h5"))
import joblib
joblib.dump(scaler, os.path.join(MODEL_DIR, "scaler.joblib"))

print("Models and scaler saved in", MODEL_DIR)


```
